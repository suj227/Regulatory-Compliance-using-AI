```python
import pandas as pd
import os

# Paths
RAW_PATH = "../data/raw/"
PROCESSED_PATH = "../data/processed/"
os.makedirs(PROCESSED_PATH, exist_ok=True)

# Load data
tx = pd.read_csv(RAW_PATH + "train_transaction.csv")
idn = pd.read_csv(RAW_PATH + "train_identity.csv")

df = tx.merge(idn, on="TransactionID", how="left")
print(df.shape)

# Basic cleaning
df["isFraud"] = df["isFraud"].astype(int)

# Drop ID-like columns
id_cols = [c for c in df.columns if c.lower().endswith("id")]
df = df.drop(columns=id_cols)

# Save cleaned data
df.to_csv(PROCESSED_PATH + "initial_cleaned_data.csv", index=False)
print("Saved initial_cleaned_data.csv")
```
