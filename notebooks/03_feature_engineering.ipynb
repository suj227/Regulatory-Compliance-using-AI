```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv("../data/processed/eda_enriched_data.csv")

# Missing values
num_cols = df.select_dtypes(include=["int64","float64"]).columns
cat_cols = df.select_dtypes(include=["object"]).columns

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[cat_cols] = df[cat_cols].fillna("Unknown")

# Features
df["log_txn_amt"] = np.log1p(df["TransactionAmt"])
df["high_value_txn"] = (df["TransactionAmt"] > df["TransactionAmt"].quantile(0.99)).astype(int)

df["txn_count_per_card1"] = df.groupby("card1")["TransactionAmt"].transform("count")
df["avg_amt_per_card1"] = df.groupby("card1")["TransactionAmt"].transform("mean")
df["amt_vs_avg"] = df["TransactionAmt"] / (df["avg_amt_per_card1"] + 1)

# Encode categoricals
for c in cat_cols:
    df[c] = LabelEncoder().fit_transform(df[c])

df.to_csv("../data/processed/feature_engineered_data.csv", index=False)
print("Saved feature_engineered_data.csv")
```
