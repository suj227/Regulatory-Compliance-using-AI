```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("../data/processed/feature_engineered_data.csv")
X = df.drop(columns=["isFraud"])
y = df["isFraud"]

X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    min_samples_leaf=50,
    class_weight="balanced",
    random_state=42
)
rf.fit(X_tr, y_tr)

train_auc = roc_auc_score(y_tr, rf.predict_proba(X_tr)[:,1])
test_auc = roc_auc_score(y_te, rf.predict_proba(X_te)[:,1])

print("Train AUC:", train_auc)
print("Test AUC:", test_auc)

summary = pd.DataFrame([{
    "train_auc": train_auc,
    "test_auc": test_auc,
    "auc_gap": abs(train_auc - test_auc)
}])

summary.to_csv("../data/processed/model_validation_summary.csv", index=False)
print("Saved model_validation_summary.csv")
```
